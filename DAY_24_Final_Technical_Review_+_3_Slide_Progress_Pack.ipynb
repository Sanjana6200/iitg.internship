```python
prs = Presentation()

prs.slide_width = Inches(13.333)
prs.slide_height = Inches(7.5)

print("Presentation initialized.")
```

```python
def add_title(
    slide,
    title
):

    title_box = slide.shapes.add_textbox(
        Inches(0.6),
        Inches(0.35),
        Inches(12),
        Inches(0.7)
    )

    text_frame = title_box.text_frame

    p = text_frame.paragraphs[0]

    p.text = title

    p.font.size = Pt(30)
    p.font.bold = True

    return title_box
```

```python
slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_title(
    slide,
    "Underwater Image Enhancement — Project Overview"
)

textbox = slide.shapes.add_textbox(
    Inches(0.8),
    Inches(1.4),
    Inches(11.8),
    Inches(5.2)
)

tf = textbox.text_frame

content = [
    "Problem: Enhance underwater images using a deep-learning image-to-image model.",
    f"Dataset: {results['dataset']}",
    f"Test samples: {results['test_samples']}",
    "Data split: Frozen train / validation / test split",
    "Preprocessing: RGB conversion, 256×256 resizing, normalization to [0,1]",
    "Architecture: U-Net",
    "Output: Enhanced RGB image"
]

for i, text in enumerate(content):

    if i == 0:
        p = tf.paragraphs[0]
    else:
        p = tf.add_paragraph()

    p.text = "• " + text
    p.font.size = Pt(21)
    p.space_after = Pt(12)
```

```python
slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_title(
    slide,
    "Model Architecture & Training"
)

textbox = slide.shapes.add_textbox(
    Inches(0.8),
    Inches(1.4),
    Inches(11.8),
    Inches(5.2)
)

tf = textbox.text_frame

content = [
    f"Architecture: {results['model']}",
    f"Loss function: {final_config['loss_function']}",
    f"Optimizer: {final_config['optimizer']}",
    f"Selected learning rate: {results['learning_rate']}",
    f"Training epochs: {final_config['training_epochs']}",
    f"Total parameters: {results['total_parameters']:,}",
    "Model status: Frozen before final test evaluation"
]

for i, text in enumerate(content):

    if i == 0:
        p = tf.paragraphs[0]
    else:
        p = tf.add_paragraph()

    p.text = "• " + text
    p.font.size = Pt(21)
    p.space_after = Pt(12)
```

```python
slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_title(
    slide,
    "Final Held-Out Test Results"
)

textbox = slide.shapes.add_textbox(
    Inches(0.8),
    Inches(1.35),
    Inches(11.8),
    Inches(5.5)
)

tf = textbox.text_frame

content = [
    f"Test L1 Loss: {results['test_L1_loss']:.6f}",
    f"Mean PSNR: {results['mean_PSNR_dB']:.4f} dB",
    f"Mean SSIM: {results['mean_SSIM']:.4f}",
    f"Average inference time: {results['average_time_per_image_seconds']:.6f} s/image",
    f"Throughput: {results['images_per_second']:.2f} images/s",
    f"Model size: {results['model_size_MB']:.4f} MB",
    f"Test samples evaluated: {results['test_samples']}"
]

for i, text in enumerate(content):

    if i == 0:
        p = tf.paragraphs[0]
    else:
        p = tf.add_paragraph()

    p.text = "• " + text
    p.font.size = Pt(21)
    p.space_after = Pt(12)
```
