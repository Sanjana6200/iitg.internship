```python
from pathlib import Path
import pandas as pd

PROJECT_DIR = Path("/content/drive/MyDrive/Underwater-Image-Data-set-main")

files_to_load = {
    "summary": "dataset_quality_summary.csv",
    "incomplete": "incomplete_triplets.csv",
    "corrupt": "corrupt_images.csv",
    "resolution": "resolution_problems.csv",
    "duplicates": "exact_duplicates.csv",
    "dark": "dark_images.csv",
    "blurry": "blurry_images.csv"
}

day1_data = {}

for key, filename in files_to_load.items():

    matches = list(PROJECT_DIR.rglob(filename))

    print(f"\n{filename}")
    print("Found:", len(matches), "file(s)")

    if len(matches) == 0:
        print(" File not found")
        day1_data[key] = pd.DataFrame()
        continue

    # Prefer Day_1 file if it exists
    day1_file = next(
        (m for m in matches if "Day_1" in str(m)),
        matches[0]
    )

    print("Using:", day1_file)
    print("Size:", day1_file.stat().st_size, "bytes")

    try:
        day1_data[key] = pd.read_csv(day1_file)
        print("✓ Loaded successfully")
        print("Shape:", day1_data[key].shape)

    except pd.errors.EmptyDataError:
        print("CSV is empty — this is okay if there were no such problems.")
        day1_data[key] = pd.DataFrame()

print("\nLOADING COMPLETE")
```

```python
EXTRACT_DIR = extracted_folders[0]

print("Using:")
print(EXTRACT_DIR)
```

```python
IMAGE_EXTENSIONS = {
    ".png",
    ".jpg",
    ".jpeg",
    ".bmp",
    ".webp"
}

image_files = [
    p
    for p in EXTRACT_DIR.rglob("*")
    if p.is_file()
    and p.suffix.lower() in IMAGE_EXTENSIONS
]

print("Total images found:", len(image_files))
```

```python
import re

pattern = re.compile(
    r"img_(\d+)_(input|target|gen)",
    re.IGNORECASE
)

records = []

for path in image_files:

    match = pattern.fullmatch(path.stem)

    if match:

        image_id = match.group(1)
        category = match.group(2).lower()

        relative_parts = path.relative_to(
            EXTRACT_DIR
        ).parts

        source_archive = relative_parts[0]

        records.append({
            "file": str(path),
            "filename": path.name,
            "source_archive": source_archive,
            "image_id": image_id,
            "category": category
        })

df = pd.DataFrame(records)

print("Images successfully identified:", len(df))

print("\nCategory distribution:")
display(df["category"].value_counts().to_frame("count"))
```

```python
sample_table = (
    df
    .pivot_table(
        index=["source_archive", "image_id"],
        columns="category",
        values="file",
        aggfunc="first"
    )
    .reset_index()
)

for column in ["input", "target", "gen"]:

    if column not in sample_table.columns:
        sample_table[column] = None

sample_table["complete"] = (
    sample_table["input"].notna() &
    sample_table["target"].notna() &
    sample_table["gen"].notna()
)

print("Total samples:", len(sample_table))

print(
    "Complete samples:",
    sample_table["complete"].sum()
)

print(
    "Incomplete samples:",
    (~sample_table["complete"]).sum()
)

display(sample_table.head())
```

```python
sample_table.to_csv(
    RESULTS_DIR / "day2_sample_pairing_table.csv",
    index=False
)

print(
    "Saved:",
    RESULTS_DIR / "day2_sample_pairing_table.csv"
)
```

```python
pairing_findings = {
    "Total samples": len(sample_table),
    "Complete input-target-gen triplets": int(
        sample_table["complete"].sum()
    ),
    "Incomplete triplets": int(
        (~sample_table["complete"]).sum()
    ),
    "Samples missing input": int(
        sample_table["input"].isna().sum()
    ),
    "Samples missing target": int(
        sample_table["target"].isna().sum()
    ),
    "Samples missing gen": int(
        sample_table["gen"].isna().sum()
    )
}

pairing_df = pd.DataFrame(
    pairing_findings.items(),
    columns=["Finding", "Count"]
)

display(pairing_df)

pairing_df.to_csv(
    RESULTS_DIR / "pairing_findings.csv",
    index=False
)
```

```python
from tqdm import tqdm
```

```python
from PIL import Image

resolution_records = []

for path in tqdm(image_files):

    try:

        with Image.open(path) as img:

            width, height = img.size

            resolution_records.append({

                "file": str(path),
                "width": width,
                "height": height,
                "resolution": f"{width}x{height}"

            })

    except:
        pass

resolution_df = pd.DataFrame(
    resolution_records
)

resolution_df = resolution_df.merge(
    df[
        [
            "file",
            "source_archive",
            "image_id",
            "category"
        ]
    ],
    on="file",
    how="left"
)

display(
    resolution_df[
        "resolution"
    ].value_counts().head(20)
)
```

```python
resolution_problems = []

for (archive, image_id), group in resolution_df.groupby(
    ["source_archive", "image_id"]
):

    if len(group) != 3:
        continue

    resolutions = set(
        zip(
            group["width"],
            group["height"]
        )
    )

    if len(resolutions) > 1:

        resolution_problems.append({

            "source_archive": archive,
            "image_id": image_id,
            "resolutions": str(resolutions)

        })

resolution_problems_df = pd.DataFrame(
    resolution_problems
)

print(
    "Triplets with inconsistent resolution:",
    len(resolution_problems_df)
)
```

```python
import cv2
```

```python
quality_records = []

for path in tqdm(image_files):

    img = cv2.imread(str(path))

    if img is None:
        continue

    gray = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2GRAY
    )

    brightness = float(gray.mean())

    contrast = float(gray.std())

    sharpness = float(
        cv2.Laplacian(
            gray,
            cv2.CV_64F
        ).var()
    )

    quality_records.append({

        "file": str(path),
        "brightness": brightness,
        "contrast": contrast,
        "sharpness": sharpness

    })

quality_df = pd.DataFrame(
    quality_records
)

quality_df = quality_df.merge(
    df[
        [
            "file",
            "source_archive",
            "image_id",
            "category"
        ]
    ],
    on="file",
    how="left"
)

display(quality_df.head())
```

```python
# Lower brightness = potentially difficult
brightness_score = (
    1 -
    quality_df["brightness"] /
    quality_df["brightness"].max()
)

# Lower sharpness = potentially difficult
sharpness_score = (
    1 -
    quality_df["sharpness"] /
    quality_df["sharpness"].max()
)

quality_df["difficulty_score"] = (
    brightness_score +
    sharpness_score
) / 2
```

```python
difficult_images = quality_df.sort_values(
    "difficulty_score",
    ascending=False
)

display(
    difficult_images[
        [
            "source_archive",
            "image_id",
            "category",
            "brightness",
            "contrast",
            "sharpness",
            "difficulty_score"
        ]
    ].head(20)
)
```

```python
difficult_images.to_csv(
    RESULTS_DIR / "difficult_samples.csv",
    index=False
)
```

```python
difficult_triplets = (
    difficult_images
    .groupby(
        ["source_archive", "image_id"]
    )["difficulty_score"]
    .mean()
    .sort_values(
        ascending=False
    )
    .head(10)
)

display(difficult_triplets)
```

```python
from sklearn.model_selection import train_test_split

complete_samples = sample_table[
    sample_table["complete"]
].copy()

print(
    "Complete samples available for splitting:",
    len(complete_samples)
)

train, temp = train_test_split(
    complete_samples,
    test_size=0.30,
    random_state=42
)

validation, test = train_test_split(
    temp,
    test_size=0.50,
    random_state=42
)

train["split"] = "train"
validation["split"] = "validation"
test["split"] = "test"

final_split = pd.concat(
    [
        train,
        validation,
        test
    ],
    ignore_index=True
)

print("Train:", len(train))
print("Validation:", len(validation))
print("Test:", len(test))
```

```python
print(
    final_split["split"]
    .value_counts()
)

print("\nPercentages:")

print(
    final_split["split"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
split_counts = (
    final_split
    .groupby(
        ["source_archive", "image_id"]
    )["split"]
    .nunique()
)

leakage_samples = split_counts[
    split_counts > 1
]

print(
    "Samples appearing in multiple splits:",
    len(leakage_samples)
)
```

```python
split_path = (
    RESULTS_DIR /
    "FROZEN_DATASET_SPLIT.csv"
)

final_split.to_csv(
    split_path,
    index=False
)

print("Frozen split saved to:")
print(split_path)
```
